```python
pd.read_sql_query("select min(order_total) as MininumTotal, max(order_total) as MaxTotal from food_delivery_orders where order_status='Completed';",conn)
```

```python
pd.read_sql_query("""select city,count(*) as Total_Orders from food_delivery_orders 
                  group by city 
                  order by Total_Orders desc;""",conn)
```

```python
pd.read_sql_query("""select customer_type,count(*) as Total_Orders from food_delivery_orders 
                  group by customer_type 
                  order by Total_Orders desc;""",conn)
```

```python
pd.read_sql_query("""select customer_type,sum(order_total) as Revenue from food_delivery_orders 
                  where order_status='Completed'
                  group by customer_type 
                  order by Revenue desc;""",conn)
```

```python
pd.read_sql_query(
    """
    SELECT
        restaurant_primary_category,
        SUM(order_total) AS Revenue
    FROM food_delivery_orders
    WHERE order_status = 'Completed'
    GROUP BY restaurant_primary_category
    ORDER BY Revenue DESC;
    """,
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT
        restaurant_primary_category,
        COUNT(*) AS TotalOrders
    FROM food_delivery_orders
    WHERE order_status = 'Completed'
    GROUP BY restaurant_primary_category
    ORDER BY TotalOrders DESC;
    """,
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT
        AVG(actual_delivery_time_minutes) AS AverageDeliveryTime
    FROM food_delivery_orders
    WHERE order_status = 'Completed';
    """,
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT
        SUM(CASE WHEN late_delivery = 1 THEN 1 ELSE 0 END) * 100.0
        / COUNT(*) AS LateDeliveryPercentage
    FROM food_delivery_orders
    WHERE order_status = 'Completed';
    """,
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT
        late_delivery,
        COUNT(*) AS TotalOrders
    FROM food_delivery_orders
    WHERE order_status = 'Completed'
    GROUP BY late_delivery
    ORDER BY late_delivery;
    """,
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT
        traffic_level,
        AVG(actual_delivery_time_minutes) AS AverageDeliveryTime
    FROM food_delivery_orders
    WHERE order_status = 'Completed'
    GROUP BY traffic_level
    ORDER BY AverageDeliveryTime DESC;
    """,
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT
        weather,
        AVG(actual_delivery_time_minutes) AS AverageDeliveryTime
    FROM food_delivery_orders
    WHERE order_status = 'Completed'
    GROUP BY weather
    ORDER BY AverageDeliveryTime DESC;
    """,
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT
        cancellation_reason,
        COUNT(*) AS CancelledOrders
    FROM food_delivery_orders
    WHERE order_status = 'Cancelled'
    GROUP BY cancellation_reason
    ORDER BY CancelledOrders DESC;
    """,
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT
        COUNT(CASE WHEN order_status = 'Cancelled' THEN 1 END) * 100.0
        / COUNT(*) AS CancellationRate
    FROM food_delivery_orders;
    """,
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT
        AVG(customer_rating) AS AverageCustomerRating
    FROM food_delivery_orders
    WHERE order_status = 'Completed';
    """,
    conn
)
```
